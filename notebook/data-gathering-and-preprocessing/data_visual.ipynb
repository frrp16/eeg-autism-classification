```python
import os
import math
import random
import numpy as np
import shutil
import pandas as pd
import matplotlib.pyplot as plt

SAMPLING_FREQ = 256
SEGMENT_LENGTH = 60
```

```python
def get_csv_EEG(filename):
    # Load data from CSV
    df_data = pd.read_csv(filename)
    return df_data, df_data.columns

def get_dummy(directory):
    files = os.listdir(directory)
    # Take random sample file in the folder
    rel_path = os.path.join(directory, random.choice(files))
    print(rel_path)
    df_data, channel_name = get_csv_EEG(rel_path)
    return df_data


```

```python
directory_source = f"datasets/grouped_csv"
```

```python
for folder in os.listdir(directory_source):
    for name in os.listdir(os.path.join(directory_source, folder)):
        df_sample = get_dummy(os.path.join(directory_source, folder, name))
        df_sample.plot(title=os.path.join(folder, name), figsize=(20,9))
```
